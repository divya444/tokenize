```python
from textblob import TextBlob
```

```python
words = ["Machine","Learnin"]
```

```python
corrected_words =[]
```

```python
for i in words:
    corrected_words.append(TextBlob(i))
```

```python
print("Wrong words: ", words)
print("Corrected words are :")
for i in corrected_words:
    print(i.correct(), end=" ")
```
